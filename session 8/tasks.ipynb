```python
"""
2.
Given a list of Flipkart product reviews with some duplicate entries, use 
value_counts() in pandas to identify which review texts are repeated most 
often and display the top 3 most common duplicate reviews."""


import pandas as pd

reviews = [
    "Good product, value for money",
    "Excellent quality",
    "Good product, value for money",
    "Fast delivery",
    "Excellent quality",
    "Good product, value for money",
    "Average product",
    "Excellent quality",
    "Fast delivery",
    "Good product, value for money",
    "Excellent quality",
    "Worth buying"
]

df = pd.DataFrame({"review": reviews})

review_counts = df["review"].value_counts()

print(review_counts.head(3))
```

```python
"""
3.
Create a DataFrame with mock data for Spotify playlists, including playlist 
names and creator usernames, where some rows are exact duplicates. Use
 drop_duplicates() to remove duplicate playlists and print the cleaned DataFrame."""


import pandas as pd

data = {
    "playlist_name": [
        "Chill Vibes",
        "Workout Hits",
        "Romantic Songs",
        "Chill Vibes",
        "Party Mix",
        "Workout Hits",
        "Lo-Fi Beats"
    ],
    "creator_username": [
        "Amit",
        "Rahul",
        "Priya",
        "Amit",
        "Neha",
        "Rahul",
        "Karan"
    ]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

cleaned_df = df.drop_duplicates()

print("\nCleaned DataFrame:")
print(cleaned_df)


```

```python
"""
4.
Suppose you have a DataFrame of Instagram usernames where some entries have typos 
(like 'insta_queen', 'insta-queen', 'instaqueen'). Use the replace() function to
 standardize all these variants to 'instaqueen'."""


import pandas as pd

data = {
    "username": [
        "insta_queen",
        "insta-queen",
        "instaqueen",
        "tech_guru",
        "insta_queen",
        "insta-queen"
    ]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

df["username"] = df["username"].replace({
    "insta_queen": "instaqueen",
    "insta-queen": "instaqueen"
})

print("\nCleaned DataFrame:")
print(df)
```

```python
"""5.
You have a DataFrame column for payment status from a Paytm-like app with mixed 
values: 'Yes', 'yes', 'Y', 'No', 'no', 'N', and some with extra spaces.
Write code to unify this column so all paid statuses become 1 and all unpaid
statuses become 0, trimming whitespace and fixing capitalization where needed.
<br><br><em><strong>Hint:</strong> Use str.strip(), str.lower(), and map/replace
methods in pandas.</em>"""


import pandas as pd

data = {
    "payment_status": [
        "Yes", "yes", "Y", "No", "no", "N",
        " Yes ", " no ", "  Y  ", " N "
    ]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

df["payment_status"] = df["payment_status"].str.strip().str.lower()

df["payment_status"] = df["payment_status"].map({
    "yes": 1,
    "y": 1,
    "no": 0,
    "n": 0
})

print("\nCleaned DataFrame:")
print(df)
```
