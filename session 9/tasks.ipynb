```python
"""
1.
Given a list of delivery timestamps as strings (e.g., ['2024-06-01 14:30', '2024-06-02 09:15'
, '2024-06-03 20:45']), use pandas and pd.to_datetime() to convert them into datetime
 objects and print the result."""

import pandas as pd

timestamps = [
    '2024-06-01 14:30',
    '2024-06-02 09:15',
    '2024-06-03 20:45'
]

dates = pd.to_datetime(timestamps)

print(dates)
```

```python
"""2.
Load a CSV file containing order dates from a Flipkart-style order history (column: 'order_date'
, format: 'YYYY-MM-DD HH:MM:SS'). Extract the year, month, and weekday for each order and add 
them as new columns in the DataFrame."""


import pandas as pd

df = pd.read_csv(r"C:\Users\HP\Downloads\flipkart_orders.csv")

df["order_date"] = pd.to_datetime(df["order_date"], format="%Y-%m-%d %H:%M:%S")

df["year"] = df["order_date"].dt.year
df["month"] = df["order_date"].dt.month
df["weekday"] = df["order_date"].dt.day_name()

print(df[["order_date", "year", "month", "weekday"]].head())
```

```python
"""3.
Set the 'order_date' column as the index of your DataFrame and use resampling to calculate the 
total number of orders placed each week.<br><br><em><strong>Hint:</strong> Use df.resample('W').size()
 after setting the datetime index.</em>"""

import pandas as pd

df = pd.read_csv(r"C:\Users\HP\Downloads\orders (1).csv", parse_dates=["order_date"])
df = df.set_index("order_date")

weekly_orders = df.resample("W").size()
print(weekly_orders)

```

```python
"""
4.
Suppose you have a DataFrame of Instagram posts with a 'posted_at' column in UTC. Convert these
 timestamps to 'Asia/Kolkata' timezone and display the first 5 converted times."""

import pandas as pd

# Sample Instagram posts (posted_at UTC me hai)
df = pd.DataFrame({
    "post_id": [5001, 5002, 5003, 5004, 5005, 5006, 5007, 5008],
    "posted_at": pd.to_datetime([
        "2024-03-01 03:47:27",
        "2024-03-02 01:41:23",
        "2024-03-02 07:38:17",
        "2024-03-02 15:58:57",
        "2024-03-04 10:35:37",
        "2024-03-05 18:20:10",
        "2024-03-06 22:45:00",
        "2024-03-08 05:10:45",
    ]),
    "likes": [1520, 980, 2310, 450, 3890, 760, 2045, 1330],
})


df["posted_at"] = df["posted_at"].dt.tz_localize("UTC")

df["posted_at_ist"] = df["posted_at"].dt.tz_convert("Asia/Kolkata")

print(df["posted_at_ist"].head())
```

```python
"""
5.
Create a new feature called 'is_weekend' in your DataFrame that marks True if an order was placed 
on Saturday or Sunday, and False otherwise.<br><br><em><strong>Constraint:</strong> Do not use any
 external libraries except pandas and numpy.</em>"""

import pandas as pd


df = pd.DataFrame({
    "order_id": [1001, 1002, 1003, 1004, 1005, 1006, 1007],
    "order_date": pd.to_datetime([
        "2024-03-01",  # Friday
        "2024-03-02",  # Saturday
        "2024-03-03",  # Sunday
        "2024-03-04",  # Monday
        "2024-03-05",  # Tuesday
        "2024-03-09",  # Saturday
        "2024-03-10",  # Sunday
    ]),
    "amount": [1200, 850, 2300, 400, 1750, 990, 3100],
})


df["is_weekend"] = df["order_date"].dt.dayofweek >= 5

df["day_name"] = df["order_date"].dt.day_name()

print(df[["order_date", "day_name", "is_weekend"]])
```
